With the Dataloop Python SDK, you'll have full control over your entire AI development lifecycle:

**Core Resources**
- **Projects** - Your high-level workspaces
- **Datasets** - Your data collections

**Advanced Features**
- **Model Management** - Train, deploy, and monitor ML models
- **Pipelines** - Build automated AI workflows
- **FaaS** - Deploy serverless functions

## 1. Authentication

### 1.1 Test Your Installation

```python
import dtlpy as dl
print(f"Dataloop SDK version: {dl.__version__}")
```

### 1.2 Load Your API Key

```python
from dotenv import load_dotenv
import os

# Load environment variables from .env file
load_dotenv()

# Access your API key securely
api_key = os.getenv('DTLPY_API_KEY')

if not api_key:
    raise ValueError("DTLPY_API_KEY not found in environment variables. Please check your .env file.")

print(f"API key loaded successfully: {api_key[:20]}...")
```

### 1.3 Initialize Dataloop

```python
# Initialize Dataloop with the API key
dl.login_api_key(api_key=api_key)

print("Logged in successfully")
```

## 2. Configuration Setup

```python
# ============================================================================
# Configuration
# ============================================================================
# TODO: Update these values for your environment
project_name = "my-first-project"
dataset_name = "my-training-data"
dataset_clone_name = "my-validation-data"
```

## 3. Create Your Project

### 3.1 Create or Get Existing Project

```python
# ============================================================================
# Project Setup
# ============================================================================
# Create your project (or get if it already exists)

try:
    # Try to get existing project
    project = dl.projects.get(project_name=project_name)
    print(f"Found existing project: {project_name}")
except dl.exceptions.NotFound:
    # Create project if it doesn't exist
    project = dl.projects.create(project_name=project_name)
    print(f"Created new project: {project_name}")
except Exception as e:
    print(f"Error creating project: {e}")
    print("Please check your permissions and try again.")
```

```python
# Verify project creation
print(f"Project ID: {project.id}")
print(f"Project Name: {project.name}")

# Open project in web browser (optional)
# project.open_in_web()
```

## 4. Create Datasets

### 4.1 Create Training Dataset

```python
# ============================================================================
# Dataset Setup
# ============================================================================
# Create your training dataset (or get if it already exists)

try:
    # Try to get existing dataset
    dataset = project.datasets.get(dataset_name=dataset_name)
    print(f"Found existing dataset: {dataset_name}")
except dl.exceptions.NotFound:
    # Create dataset if it doesn't exist
    dataset = project.datasets.create(dataset_name=dataset_name)
    print(f"Created new dataset: {dataset_name}")
except Exception as e:
    print(f"Error creating dataset: {e}")
```

### 4.2 Clone Dataset for Validation

```python
# Clone the training dataset to create a validation dataset
try:
    cloned_dataset = dataset.clone(clone_name=dataset_clone_name)
    print(f"Created cloned dataset: {dataset_clone_name}")
except Exception as e:
    print(f"Error cloning dataset: {e}")
```

### 4.3 List and Verify Datasets

```python
# List all datasets in the project
datasets = project.datasets.list()
print(f"Total datasets in project: {len(datasets)}")
for ds in datasets:
    print(f"  - {ds.name}")
```

## 6. Best Practices

### Project Organization
- Use clear, descriptive naming conventions for projects and datasets
- Add detailed descriptions to your resources for better documentation
- Maintain consistent structure across your projects
- Tag resources appropriately for easy filtering and organization

### Security Practices
- Always use environment variables for sensitive data like API keys
- Never hardcode credentials in your code
- Regularly rotate your API keys
- Use appropriate role-based access control for team members

### Resource Management
- Clean up unused test datasets and projects regularly
- Use appropriate dataset naming to avoid conflicts
- Monitor your storage usage and optimize as needed

## 7. Troubleshooting Guide

### Common Issues

#### Authentication Failures

```python
# Check token status
if dl.token_expired():
    print("Token is expired. Please re-authenticate.")
    dl.login_api_key(api_key=api_key)
else:
    print("Token is valid.")
```

#### Project Access Issues

```python
# Verify project existence
try:
    project = dl.projects.get(project_name=project_name)
    print(f"Project found: {project.name}")
except dl.exceptions.NotFound:
    print("Project not found. Please check the project name.")
except Exception as e:
    print(f"Error accessing project: {e}")
```

#### Dataset Operations

```python
# Handle dataset errors
try:
    dataset = project.datasets.get(dataset_name=dataset_name)
    print(f"Dataset found: {dataset.name}")
except dl.exceptions.NotFound:
    print("Dataset not found. Please check the dataset name.")
except dl.exceptions.Forbidden:
    print("Insufficient permissions to access this dataset.")
except Exception as e:
    print(f"Error accessing dataset: {e}")
```
